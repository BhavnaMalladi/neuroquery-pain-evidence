# Representation bias in the NeuroQuery pain corpus

This notebook reads the precomputed tables in `data/` and shows each result. To rebuild the tables, run the scripts in `src/` as listed in the README.

```
import os, json
import pandas as pd
from IPython.display import Image, display
DATA, FIG = 'data', 'figures'
def show(csv=None, png=None):
    if csv and os.path.exists(os.path.join(DATA, csv)):
        display(pd.read_csv(os.path.join(DATA, csv)))
    if png and os.path.exists(os.path.join(FIG, png)):
        display(Image(os.path.join(FIG, png)))
```

## A. Size-fair reporting audit

Fraction of pain studies reporting a peak near each region, with a sphere of the same radius at every region. The ratio of the cortical mean to the subcortical mean is reported across radii, with a threshold sweep to show the selection does not drive it.

```
show('ratio_by_radius.csv', 'reporting_forest.png')
show('threshold_sweep.csv', 'ratio_by_radius.png')
```

## B. Region sampling of the predictive map

NeuroQuery's encoding map for the single query 'pain', read at the region coordinates. This is a predictive encoding value, reported on its own terms.

```
show('roi_zvalues.csv', 'roi_zvalues.png')
```

## C. Mention frequency versus reporting frequency

For the same pain studies, how often each region is named in the text versus how often a coordinate is reported for it. A gap isolates coordinate reporting from discussion.

```
show('mention_vs_reporting.csv', 'mention_vs_reporting.png')
```

## D. Change across publication year

The cortical to subcortical reporting ratio by year, with a weighted trend. A falling ratio points to a measurement limit, a flat ratio to a reporting habit.

```
show('year_trend.csv', 'year_trend.png')
```

## E. Semantic anchoring

How strongly 'pain' co-occurs with cortical versus subcortical region terms across the corpus, computed from raw counts with no model involved.

```
show('semantic_anchoring.csv', 'semantic_anchoring.png')
show(png='pain_nearest_terms.png')
show('semantic_anchoring_terms.csv', 'semantic_anchoring_terms.png')
```
